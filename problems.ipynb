```python
import numpy as np
import pandas as pd
```

```python
ci = {90 : 1.645, 95 : 1.96, 99 : 2.576}
rng = np.random.default_rng(seed=100)
R = 5000
```

```python
df7 = pd.read_csv('data/ames_prices.csv')
df7.head()
```

```python
df7['indicator']= (df7['Central.Air'] == 'Y').astype(int)
prp = df7['indicator'].mean()
print(f'Proportion of homes with central air : {round(prp,2)}')
```

```python
col = df7['indicator'].to_numpy()
n = len(col)
bootstrap7 = []

for i in range(R):
    col_b = rng.choice(col, size=n, replace=True)
    mean_b7 = col_b.mean()
    bootstrap7.append(mean_b7)
```

```python
se_hat7 = np.std(bootstrap7, ddof=1)
se_hat7
```

```python
for i,j in ci.items():
    lower7 = prp - j * se_hat7
    upper7 = prp + j * se_hat7
    print(f'For {i}% confidence interval is ({round(lower7,3)},{round(upper7,3)})')

    if lower7 < 0.90 < upper7:
        print(f'For {i}% : Value within CI. Fail to reject null hypothesis')
    else:
        print(f'For {i}% : Reject null hypothesis')
```

```python
df8 = pd.read_csv('data/ames_prices.csv')
df8.head()
```

```python
area_hyp = np.arcsinh(df8['area'])
price_hyp = np.arcsinh(df8['price'])
```

```python
pear_corr = area_hyp.corr(price_hyp,'pearson')
print(f'The pearson correlation is : {round(pear_corr,3)}')
```

```python
col1 = area_hyp.to_numpy()
col2 = price_hyp.to_numpy()
bootstrap8 = []

for i in range(R):
    r = rng.integers(0,n,size=n)
    col1_b = col1[r]
    col2_b = col2[r]
    pear_corr_b = np.corrcoef(col1_b,col2_b)[0,1]
    bootstrap8.append(pear_corr_b)
```

```python
se_hat8 = np.std(bootstrap8, ddof=1)
se_hat8
```

```python
for i,j in ci.items():
    lower8 = pear_corr - j * se_hat8
    upper8 = pear_corr + j * se_hat8
    print(f'For {i}% confidence interval is ({round(lower8,3)},{round(upper8,3)})')

    if lower8 < 0 < upper8:
        print(f'For {i}% : Value within CI. Fail to reject null hypothesis')
    else:
        print(f'For {i}% : Reject null hypothesis')
```

```python
df9 = pd.read_csv('data/metabric.csv')
df9.head()
```

```python
overall_mean = df9['Overall Survival (Months)'].mean()
print(f'The estimated mean of overall survival is : {round(overall_mean,3)} months or {round(overall_mean/12,0)} years')
```

```python
mon = df9['Overall Survival (Months)'].to_numpy()
n = len(mon)
bootstrap9 = []

for i in range(R):
    mon_b = rng.choice(mon, size=n, replace=True)
    mean_b9 = mon_b.mean()
    bootstrap9.append(mean_b9)
```

```python
se_hat9 = np.std(bootstrap9, ddof=1)
se_hat9
```

```python
for i,j in ci.items():
    lower9 = overall_mean - j * se_hat9
    upper9 = overall_mean + j * se_hat9
    print(f'For {i}% confidence interval is ({round(lower9,3)},{round(upper9,3)}) in months')
    print(f'For {i}% confidence interval is ({round(lower9/12,1)},{round(upper9/12,1)}) in years')

    if lower9 < 120 < upper9:
        print(f'For {i}% : Value within CI. Fail to reject null hypothesis')
    else:
        print(f'For {i}% : Reject null hypothesis')
```

```python
df10 = pd.read_csv('data/nhanes_data_17_18.csv',low_memory=False)
df10.head()
```

```python
df10_new = df10[(df10['AgeInYearsAtScreening'] >= 18) & df10['SleepHoursWeekdaysOrWorkdays'].notna()]
sleep = df10_new['SleepHoursWeekdaysOrWorkdays'].mean()
print(f'Estimated mean sleep for people >=18 years age on weekdays = {round(sleep,3)}')
```

```python
week_sleep = df10_new['SleepHoursWeekdaysOrWorkdays'].to_numpy()
n = len(week_sleep)
bootstrap10 = []

for i in range(R):
    week_sleep_b = rng.choice(week_sleep, size=n, replace=True)
    mean_b10 = week_sleep_b.mean()
    bootstrap10.append(mean_b10)
```

```python
se_hat10 = np.std(bootstrap10, ddof=1)
se_hat10
```

```python
for i,j in ci.items():
    lower10 = sleep - j * se_hat10
    upper10 = sleep + j * se_hat10
    print(f'For {i}% confidence interval is ({round(lower10,3)},{round(upper10,3)})')

    if lower10 < 8 < upper10:
        print(f'For {i}% : Value within CI. Fail to reject null hypothesis')
    else:
        print(f'For {i}% : Reject null hypothesis')
```
